```python
import json
import pandas as pd
from collections import defaultdict
```

```python
def json_to_dict(obj):
    return json.loads(obj)

def load_csv(path):
    df = pd.read_csv(path, dtype={'username':str})
    df['counts'] = df['counts'].map(json_to_dict)
    return df
```

```python
df = load_csv(r'PRAW\count_data\all_batches_sub_posts_count.csv')
```

```python
total_counts = defaultdict(int)
for counts in df['counts']:
    for subreddit, count in counts.items():
        total_counts[subreddit] += count
sorted_counts = dict(sorted(total_counts.items(), key=lambda item: item[1], reverse=True))
```

```python
key_max = list(sorted_counts.keys())[0]
for i in range(200-14):
    key_current = list(sorted_counts.keys())[i]
    print(f"{sorted_counts[key_current]/sorted_counts[key_max]*100-1:.0f}, {key_current}")
```
