# Merges CSVs

## nullArray CSVs

*nullArray CSVs are only one column**

```python
with open('nullArray-merged.csv', 'w') as fi:
    fi.write('null_users\n')
    for i in range(9):
        with open(f'nullArray{i}.csv', 'r') as f:
            fi.write(f.read())
```

## userData CSVs

```python
import csv
with open("userData-merged.csv", "w", newline="", encoding="utf-8") as fout:
    writer = csv.writer(fout)
    # explicitly define your columns
    writer.writerow(["author","cake_day_utc","karma","post_karma","comment_karma"])
    for i in range(9):
        with open(f"userData{i}.csv", newline="", encoding="utf-8") as fin:
            reader = csv.reader(fin)
            next(reader)              # skip each file’s header
            writer.writerows(reader)  # write only the data rows
```
